## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

```python
# Section 5: Identify the signals that make the problem more complex than one fixed rule

signals = [
    "impressions",
    "clicks",
    "CTR",
    "average position",
    "search volume",
    "content age",
    "engagement",
    "recent trends"
]

print("Signals considered:")
for signal in signals:
    print("-", signal)

print("\nWhy ML:")
print("These multiple signals can be considered together instead of using one fixed threshold.")
```

### Why ML beats a fixed rule

A simple rule such as "update every page whose clicks decreased by 20%" would consider only one condition. Content performance depends on multiple factors, including impressions, clicks, CTR, average position, search volume, content age, engagement, and recent trends.

An ML approach could learn patterns from these different signals and produce a priority score or ranking. This would support the content team in deciding which pages to review or update first instead of relying on one fixed threshold.
